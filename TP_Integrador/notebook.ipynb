```python
# Corre las 9 frases con los dos modelos usando pipeline, las compara con lo esperado y arma la tabla de resultados (versión con salida limpia).

import time  # para medir cuánto tarda cada llamada
import pandas as pd  # para armar la tabla de resultados
from transformers import pipeline  # herramienta que arma el extractor de entidades
from transformers.utils import logging as hf_logging  # control de los avisos de transformers
from huggingface_hub.utils import logging as hub_logging  # control de los avisos de huggingface_hub

hf_logging.set_verbosity_error()  # oculta los avisos de transformers, deja solo los errores
hf_logging.disable_progress_bar()  # oculta la barra "Loading weights"
hub_logging.set_verbosity_error()  # oculta el aviso del token de Hugging Face

modelos = ["Babelscape/wikineural-multilingual-ner", "Davlan/xlm-roberta-base-ner-hrl"]  # los dos candidatos a comparar
filas = []  # acá se guarda una fila por cada entidad esperada
de_mas = []  # acá se guarda lo que el modelo marcó y no estaba en lo esperado
tiempos = []  # acá se guarda cuánto tardó cada llamada

for nombre in modelos:  # repite todo para cada modelo
    ner = pipeline("ner", model=nombre, aggregation_strategy="simple")  # arma el extractor de este modelo
    for numero, f in enumerate(frases, start=1):  # recorre las frases, numerándolas desde 1
        inicio = time.perf_counter()  # anota la hora justo antes de la llamada
        resultado = ner(f["texto"])  # analiza la frase: esto es la inferencia
        segundos = round(time.perf_counter() - inicio, 3)  # calcula cuánto tardó
        tiempos.append({"modelo": nombre, "frase": numero, "segundos": segundos})  # guarda ese tiempo
        detectadas = [(e["word"].strip().lower(), e["entity_group"], float(e["score"])) for e in resultado]  # lista simple (texto, tipo, puntaje) de lo detectado
        for tipo_esp, texto_esp in f["esperado"]:  # revisa cada entidad esperada de esta frase
            tipo_det, puntaje = "no detectada", None  # por defecto: el modelo no la encontró
            for palabra, grupo, score in detectadas:  # busca entre lo detectado
                if palabra == texto_esp.lower() and (tipo_det == "no detectada" or grupo == tipo_esp):  # mismo texto, y se prefiere el mismo tipo
                    tipo_det, puntaje = grupo, round(score, 3)  # anota el tipo y el puntaje detectados
            filas.append({"modelo": nombre, "frase": numero, "tipo_frase": f["tipo"], "entidad": texto_esp, "tipo_esperado": tipo_esp, "tipo_detectado": tipo_det, "puntaje": puntaje})  # agrega la fila a la tabla
        textos_esperados = [t.lower() for _, t in f["esperado"]]  # textos esperados, en minúscula
        for palabra, grupo, score in detectadas:  # revisa lo que el modelo marcó
            if palabra not in textos_esperados:  # si no estaba entre lo esperado
                de_mas.append({"modelo": nombre, "frase": numero, "texto": palabra, "tipo": grupo, "puntaje": round(score, 3)})  # lo guarda aparte

tabla = pd.DataFrame(filas)  # convierte las filas en una tabla
tabla["acerto"] = (tabla["tipo_esperado"] == tabla["tipo_detectado"]).map({True: "sí", False: "no"})  # "sí" si el tipo detectado coincide con el esperado
pd.set_option("display.max_rows", None)  # muestra todas las filas, sin cortar

for nombre, grupo in tabla.groupby("modelo"):  # un resumen por modelo
    print(nombre, ":", (grupo["acerto"] == "sí").sum(), "de", len(grupo), "entidades acertadas")  # muestra los aciertos en un renglón

tabla  # muestra la tabla completa
```

```python
# Calcula el tiempo promedio y el máximo por llamada de cada modelo con pipeline.
pd.DataFrame(tiempos).groupby("modelo")["segundos"].agg(["mean", "max"]).round(3)
```

```python
# Pide el token por teclado para usarlo sin dejarlo escrito en la notebook, y arma el cliente de InferenceClient.

from getpass import getpass  # función que pide un texto sin mostrarlo ni guardarlo en el archivo
from huggingface_hub import InferenceClient  # el cliente que llama a los modelos en los servidores de Hugging Face

token = getpass("Pegá tu token de Hugging Face: ")  # en VS Code aparece una cajita arriba de la ventana
client = InferenceClient(provider="hf-inference", api_key=token)  # arma el cliente con tu token
```

```python
# Hace una llamada de prueba con InferenceClient y muestra qué tipo de objeto devuelve.

frase = "Lionel Messi nació en Rosario y juega en el Inter Miami."  # la misma frase de la primera prueba con pipeline
resultado_ic = client.token_classification(frase, model="Babelscape/wikineural-multilingual-ner")  # la llamada viaja por internet y corre en los servidores de Hugging Face
print(resultado_ic)  # muestra lo que devolvió
print(type(resultado_ic))  # tipo del resultado completo
print(type(resultado_ic[0]))  # tipo de cada elemento
```
